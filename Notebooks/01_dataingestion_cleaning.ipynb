## Joining the client and ticket table

```python
tickets_clients = df_tickets.merge(df_clients, on="ClientID", how="left")
```

```python
final_df = tickets_clients.merge( df_agents,left_on="AssignedAgentID",right_on="AgentID",how="left",
    suffixes=("", "_Agent"))

final_df.head()
```

```python
print(final_df.shape)
print(final_df.head())
print(final_df.columns.tolist())
```

```python
print(final_df["ClientName"].isna().sum())
```

```python
print("Tickets + Clients:", len(tickets_clients))
print("Final dataset:", len(final_df))
print("Missing AgentID after merge:", final_df["AgentID"].isna().sum())
```

```python
final_df["IsResolved"] = final_df["ResolvedAt"].notna().astype(int)
final_df["IsResolved"] = (final_df["ResolvedAt"].notna().astype(int))
final_df["ResolutionNotesAvailable"] = (final_df["ResolutionNotes"].notna().astype(int))
final_df["IsResolvedByStatus"] = (final_df["Status"].isin(["Closed", "Resolved"]).astype(int))
```

```python
print(final_df["FirstResponseAvailable"].value_counts())
print(final_df["IsResolved"].value_counts())
print(final_df["ResolutionNotesAvailable"].value_counts())
print(final_df["IsResolvedByStatus"].value_counts())
```

### Converting the dates 

```python
date_columns = ["CreatedAt","FirstResponseAt", "ResolvedAt"]

for col in date_columns:final_df[col] = pd.to_datetime(final_df[col], errors="coerce")
print(final_df[date_columns].dtypes)
```

### Calculate response and resolution hours

```python
final_df["FirstResponseHours"] = (final_df["FirstResponseAt"] - final_df["CreatedAt"]).dt.total_seconds() / 3600

final_df["ResolutionHours"] = ( final_df["ResolvedAt"] - final_df["CreatedAt"]).dt.total_seconds() / 3600
print(final_df[["FirstResponseHours", "ResolutionHours"]].describe())
```

```python
print(final_df.columns.tolist())
```

```python
print(final_df["SLABreached"].value_counts(dropna=False))
print(final_df[["SLADueAt", "SLABreached", "Status", "ContractTier", "PriorityID"]].head(10))
print(final_df[["SLADueAt", "SLABreached", "FirstResponseHours", "ResolutionHours"]].dtypes)
```

```python
print(final_df["SLABreached"].value_counts(dropna=False))

print( final_df["SLABreached"].value_counts(normalize=True, dropna=False).mul(100).round(2))

sla_summary = pd.DataFrame({"Count": final_df["SLABreached"].value_counts(dropna=False),"Percentage": (final_df["SLABreached"]
.value_counts(normalize=True, dropna=False).mul(100).round(2))})

print(sla_summary)
```

```python
priority_sla = ( final_df.groupby("PriorityID")["SLABreached"].agg(["count", "sum", "mean"]).reset_index())

priority_sla["BreachPercentage"] = (priority_sla["mean"] * 100).round(2)

priority_sla = priority_sla.rename(columns={"count": "TotalTickets","sum": "BreachedTickets"})

priority_sla = priority_sla[["PriorityID", "TotalTickets", "BreachedTickets", "BreachPercentage"]]

print(priority_sla)
```
